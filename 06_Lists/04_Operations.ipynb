__Operations on list__

Arithmatic (+,*)

```python
l1 = [1,2,3,4]
l2 = [5,6,[7,8]]

print(l1 + l2)
print(l1*3)
```

Membership operations (in, not in)

```python
l1 = [1,2,3,4]
l2 = [5,6,[7,8]]

print(7 in l2)
print([7,8] in l2)


print(5 not in l1)
print(4 not in l1)
```

Loops

```python
l1 = [1,2,3,4]
l2 = [5,6,[7,8]]

for i in l1:
    print(i)
```

```python
for i in l2:
    print(i)
```
