Two ways to traverse a list

itemwise

```python
l= [1,2,3,4]

for i in l:
    print(i)
```

indexwise

```python
for i in range(0,len(l)):
    print(l[i])
```

write a program to add items of 2 lists indexwise.

```python
l1 = [1,2,3,4]
l2 = [-1,-2,-3,-4]
list(zip(l1,l2))
[i + j for i,j in zip(l1,l2)]
```
