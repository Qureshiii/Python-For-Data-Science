__append__

```python
l= [1,2,3,4,5]
l.append(8)
l.append(True)
l.append([9,10,11]) # this is when you add a multiple items in a list using append function it will add the entire list not to broke in pieces and then add
print(l)
```

__Extend__

```python
l= [1,2,3,4,5]
l.extend([6,7,9])
l.extend('Bannu') # this is when you add a string. the extend function broke it in pieces and add in a list.
print(l)
```

__Insert__

```python
l= [1,2,3,4,5]
l.insert(1,100)
l.insert(5,True)
print(l)
```
