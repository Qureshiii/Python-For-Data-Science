__Editting items in a list__

```python
l= [1,2,3,4,5]
l[0] = 500
l[-1] = True
l[1:4] = 200,300,400
print(l)
```

__Deleting items from a list__

del / here we can delete multiple items from a list using index positions

```python
l= [1,2,3,4,5]
del l[-1]
print(l)
del l[1:3]
print(l)
```

remove / here we can delete items from a list using its value.

```python
l= [1,2,3,4,5]
l.remove(5)
print(l)
```

pop

```python
l= [1,2,3,4,5]
l.pop(0)
l.pop()
print(l)
```

clear

```python
l= [1,2,3,4,5]
l.clear()
print(l)
```
