__List Comprehension from Basic to Advance__

The traditional Long way.

```python
number = []

for i in range(1,6):
    number.append(i)

print(number)
```

The magic shortcut way.

```python
numbers = [i for i in range(1,6)]
print(numbers)
```

Doing math and changing things

```python
double = []
for i in range(1,6):
    double.append(i*2)
print(double)

# Or

doubled = [x*2 for x in range(1,6)]
print(doubled)
```

Capitalize the first latter.

```python
characters = ['mario','luigi','bowser']
capital = [name.capitalize() for name in characters]
print(capital)
```

```python
# The long way

evens = []
for i in range(1,11):
    if i % 2 == 0:
        evens.append(i)
print(evens)
        
        
        #Or The shortcut way
        
even = [x for x in range(1,11) if x % 2 == 0 ]
print(even)        
```

Extract only odd numbers

```python
numbers = [12,15,22,27,30,33]
#Extract only the odd numbers
odds = [ num for num in numbers if num % 2 != 0]
print(odds)
```

handling output with if-else 

```python
# if else 
scores = [45,82,67,50,95]
#Label exam score as 'pass' and 'fail'
results = ['Pass' if i >=60 else 'Fail' for i in scores ]
print(results)
```

Text Analyzer / Spam filter (extract all words that are longer than 4 letters)

```python
sentence = 'Python is an amazing language to learn'
words = sentence.split()
long_words = [i for i in words if len(i) > 4]
print(long_words)
```

User Name Cleaner / Scenario: users are creating accounts on your app. some type their names in lowercase, some in uppercase, and some put weird spaces. you want to normalize them by making them lowercase and stripping spaces, but you completely skip anyone who leaves it blank.

```python
raw_users = ['Alice','bob',' ','CHARLIE']
clean_users = [name.strip().lower() for name in raw_users if name.strip() !='']

print(clean_users)
```

Mystery

```python
numbers = [1,2,3,4,5]
mystery = [i * i for i in numbers if i % 2 != 0]
print(mystery)
```

Scalar multiplication on a vector.

```python
v = [2,3,4]
s = -3
multi = [s * i for i in v]
print(multi)
```

Squares

```python
l = [1,2,3,4,5]
squares = [i ** 2 for i in l]
print(squares)
```

Print all the numbers which are divisible by 5 in range of 1 to 50.

```python
num = [i for i in range(1,50) if i % 5 == 0]
print(num)
```

find the language which is starts with letter p.

```python
lang = ['java', 'python', 'php','c','javascript']

starts_P = [i for i in lang if i.startswith('p')]
print(starts_P)
```

```python
languages = ['java', 'python', 'php','c','javascript']
starts_j = [language for language in languages if language.startswith('j')]
print(starts_j)
```

Mix to lists together using nested if

```python
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']

[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
#[i for i in my_fruits if i in basket if fruit.startswith('a')]
```

print a (3,3) matrix using list comprehension/ nested

```python
[[i * j for i in range(1,4)] for j in range(1,4)]
```

cartesian product --> list comprehension using 2 lists together

```python
l1 = [1,2,3,4]
l2 = [5,6,7,8]

[i * j for i in l1 for j in l2]
```
