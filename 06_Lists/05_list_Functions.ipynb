len / min / max / sorted

```python
l = [2,1,5,7,0]
print(len(l))
print(min(l)) # this is used only for homogenuis data
print(max(l)) # this is used only for homogenuis data
print(sorted(l))
print(sorted(l,reverse=True))
```

count / it tell about any item frequency count in the list

```python
l = [1,2,1,3,4,1,5]
l.count(1)
l.count(3)
```

index

```python
l = [1,2,1,3,4,1,5]
l.index(5)
```

reverse / it permanently reverses the list

```python
l = [1,2,1,3,4,1,5]
l.reverse()
print(l)
```

sort vs sorted

```python
l = [2,1,5,7,0]
print(l)
print(sorted(l))
print(l)
l.sort()
print(l)
```

copy / creats a shallow copy of the given data.

```python
l = [2,1,5,7,0]
print(l)
print(id(l))
l1 = l.copy()
print(l1)
print(id(l1))
```
