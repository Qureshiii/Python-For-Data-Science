let's create a function that tells us about the given number is even.

```python
def is_even(num):
    """
    this function is return if the given number is odd or even.
    input - any valid integer
    output - odd/even
    created on 29 june 2026.
    """
    if type(num) == int:
        if num % 2 == 0:
            return 'Even'
            
        else:
            return 'Odd'
    else:
        return "this function only works on integers."


is_even('Hello')
    
```

Now calling a Function

```python
# Function_Name(input)
for i in range(1,11):
    x = is_even(i)
    print(x)
```

*args

```python
def multiply(*args):
    
    product = 1
    
    for i in args:
        product = product * i
        
    return product
```

```python
multiply(1,2,3,4,5,6)
```

```python
def g(x):
    def h():
        x = 'abc'
    x = x + 1
    print ('in g(x): x =', x)
    h()
    return x

x = 3
z = g(x)
```

Another Example

```python
def g(x):
    def h(x):
        x = x+1
        print('in h(x): x =', x)
    x = x + 1
    print('in g(x): x =', x)
    h(x)
    return x

x = 3
z = g(x)
print('in main program scope: x = ', x)
print('in main program scope: x = ', z)
```
