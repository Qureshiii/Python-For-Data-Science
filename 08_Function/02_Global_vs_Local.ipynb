__Global variable Vs Local Variable__

```python
time_of_day = 'night'# Global variable
def kitchen():
    item = 'Flashlight' # Local variable
    print(f'it is {time_of_day} and i found a {item}')
    
def library():
    #Print(item) #we can't use here the item cuz it is local variable in kitchen()
    print(f'it is {time_of_day} but this room is too dark!.')# we can use global variable everywhere
    
kitchen()
library()
```
