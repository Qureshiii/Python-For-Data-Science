Higher Order Function Example.

```python
def square(x):
    return x**2

# this transform function is actually a Higher order Function.cuz it recive function as an argument.
def transform(f,l):
    output = []
    for i in l:
        output.append(f(i))
        
    print(output)
    
l = [1,2,3,4,5,6,7,8,9]

# instead of creating multiple function we can simply use lambda function here.
print(transform(square,l))

print(transform(lambda x : x**2,l))
# for cubes
print(transform(lambda x : x**3,l))
```

The E-Commerce Shoping Cart.

```python
from functools import reduce

shopping_cart = [
    {'name':'Laptop','catagory':'electronics','price':1000},
    {'name':'T-Shirt','catagory':'clothing','price':25},
    {'name':'Headphones','catagory':'electronics','price':100},
    {'name':'Book','catagory':'books','price':15},
    
]
```

```python
#Applying filter first
electronics = list(filter(lambda x: x['catagory']=='electronics',shopping_cart))
```

```python
#Applying map 10% Discount
discounted_prices = list(map(lambda x:x['price'] * 0.9,electronics ))
```

```python
#Appling reduce 
total_cost = reduce(lambda x,y:x+y,discounted_prices)
total_cost
```

Cleaning and Normalizing API logs.

```python
raw_logs = [
    {"status": "200", "msg": "  user_login_success  "},
    {"status": "500", "msg": "internal_server_error"},
    {"status": "200", "msg": "FETCH_DATA_SUCCESS  "},
    {"status": "404", "msg": "page_not_found"}
]
```

```python
# keep only successful logs using filter function()
successful_logs = list(filter(lambda x:x['status']=='200',raw_logs))
```

```python
# clean up the whitespace and lowercase the messages using map().
cleaned_logs = list(map(lambda x:x['msg'].strip().lower(),successful_logs))
cleaned_logs
```

Extracting the country san francisco from the deep oceans.

```python
from functools import reduce

user_profile = {
    "user": {
        "demographics": {
            "location": {
                "city": "San Francisco",
                "timezone": "PST"
            }
        }
    }
}

```

```python
path = ['user','demographics','location','city']
city = reduce(lambda x,y:x.get(y,{}),path,user_profile)
```

```python
print(city)
```

__Pipeline Builder__

```python
# Define the three independence proccessing functions.
strip_spaces = lambda x: x.strip()
capitalize_all = lambda x: x.upper()
add_exclamation = lambda x: x + '!'

def compose(*functions):
    return reduce(lambda f,g: lambda x: g(f(x)),functions)

text_pipeline = compose(strip_spaces,capitalize_all,add_exclamation)

raw_strings = ["  hello  ", "  python is fun", "functional programming   "]
proccessed_strings = list(map(text_pipeline,raw_strings))

print(proccessed_strings)
```
