creating a Tuple

```python
#Empty tuple

t1 = ()
print(t1)

# Single item tuple/ we will use a comma while creating a single item tuple

t2 = (5,)
print(t2)
print(type(t2))

# Homogenuis Tuples
t3 = (1,2,3,4)
print(t3)
print(type(t3))


#Haterogenuis tuple

t4 = (1,2.3,True,[1,2,True])
print(t4)

#3d 
t5 = (1,2,3,(21,33,44))
print(t5)

t6 = tuple('hello')
print(t6)
```

Special Syntax

```python
a,b,c = (1,2,3)
print(a,b,c)
```

```python
a = 1
b = 2
a,b = b,a
print(a,b)
```

Zipping tuple

```python
t1 = (1,2,3,4)
t2 = (5,6,7,8)

tuple(zip(t1,t2))
```
