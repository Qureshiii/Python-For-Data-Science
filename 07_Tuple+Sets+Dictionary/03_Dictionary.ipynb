Dictionary Comprehension

Print first 10 numbers and their squares.

```python
{i: i ** 2 for i in range(1,11)}
```

Using existing dictionary

```python
distances = {'Bannu':1000,'Islamabad':2000,'lahore':3000}
# convert this kilometer into miles
{i:j*0.62 for i,j in distances.items()}
```

Dictionary comprehension using zip

```python
days = {'sunday','monday','tuesday','wednesday','thursday','friday','saturday'}
temp_c = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for (i,j) in zip(days,temp_c)}
```

Dictionary comprehension using if condition

```python
products = {'phone':10,'laptop':0,'charger':32,'Tablet':0}

{key:value for (key,value) in products.items() if value > 0}
```

Nested Comprehension

Print table of numbers from 2 to 4

```python
{i:{j:i*j for j in range(1,11)}for i in range(2,5)}
```
